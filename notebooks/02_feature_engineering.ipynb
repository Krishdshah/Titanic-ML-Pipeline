# Titanic Survival Prediction

## Feature Engineering

Goal:
Create meaningful features that improve model performance and capture hidden patterns in passenger data.

```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
```

```python
train = pd.read_csv("train.csv")
test = pd.read_csv("test.csv")

train.shape, test.shape
```

```python
test_ids = test["PassengerId"]
```

```python
full = pd.concat(
    [train.drop("Survived", axis=1), test],
    axis=0,
    ignore_index=True
)

full.shape
```

```python
full["Title"] = full["Name"].str.extract(
    " ([A-Za-z]+)\.",
    expand=False
)
```

```python
full["Title"].value_counts()
```

```python
rare_titles = [
    "Lady",
    "Countess",
    "Capt",
    "Col",
    "Don",
    "Dr",
    "Major",
    "Rev",
    "Sir",
    "Jonkheer",
    "Dona"
]

full["Title"] = full["Title"].replace(
    rare_titles,
    "Rare"
)

full["Title"] = full["Title"].replace({
    "Mlle": "Miss",
    "Ms": "Miss",
    "Mme": "Mrs"
})
```

```python
full["FamilySize"] = (
    full["SibSp"]
    + full["Parch"]
    + 1
)
```

```python
full["IsAlone"] = (
    full["FamilySize"] == 1
).astype(int)
```

```python
full["CabinKnown"] = (
    full["Cabin"]
    .notnull()
    .astype(int)
)
```

```python
full.groupby("Title")["Age"].median()
```

```python
full["Age"] = (
    full.groupby("Title")["Age"]
    .transform(
        lambda x: x.fillna(x.median())
    )
)
```

```python
full["Fare"] = (
    full["Fare"]
    .fillna(
        full["Fare"].median()
    )
)
```

```python
full["Embarked"] = (
    full["Embarked"]
    .fillna(
        full["Embarked"].mode()[0]
    )
)
```

```python
full["AgeBand"] = pd.cut(
    full["Age"],
    bins=5,
    labels=False
)
```

```python
full["FareBand"] = pd.qcut(
    full["Fare"],
    q=4,
    labels=False,
    duplicates="drop"
)
```

```python
ticket_counts = (
    full["Ticket"]
    .value_counts()
)
```

```python
full["TicketGroupSize"] = (
    full["Ticket"]
    .map(ticket_counts)
)
```

```python
full.head()
```

```python
full.to_csv(
    "engineered_features.csv",
    index=False
)

print("Saved successfully.")
```
